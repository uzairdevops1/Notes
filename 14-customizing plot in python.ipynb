```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pf
import numpy as np
kashti=sns.load_dataset("Titanic")
kashti.head()
```

```python
sns.boxplot(x="sex",
           y="fare", data=kashti)
```

```python

sns.boxplot(x="survived",
           y="age", showmeans=True,
           meanprops={"marker":"*",
                    "markersize":"12",
                    "markeredgecolor":"red"},
           data=kashti)
# show labels
plt.xlabel("how many survived", size = 12, weight='bold')
plt.ylabel("age (years)", size= 14, weight='bold')
plt.title("box plot of kitny doobay or kitnay bach gye",size = 16, weight='bold')

plt.show()
```
