## change color

```python
import seaborn as sns
import matplotlib.pyplot as plt
kashti=sns.load_dataset("titanic")
kashti
sns.barplot(x="sex", y="alone", hue="who", data=kashti, order=["female", "male"], color="brown")
plt.show
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
kashti=sns.load_dataset("titanic")
kashti
sns.barplot(x="sex", y="alone", hue="who", data=kashti, order=["female", "male"], color="brown", palette='pastel')
plt.show
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
kashti=sns.load_dataset("titanic")
kashti
sns.barplot(x="sex", y="alone", hue="who", data=kashti, order=["female", "male"], color="brown", palette='pastel', ci=None)
plt.show
```

```python
import seaborn as sns
from numpy import mean
import matplotlib.pyplot as plt
kashti=sns.load_dataset("titanic")
kashti
sns.barplot(x="sex", y="fare", hue="who", data=kashti, order=["female", "male"], color="brown", palette='pastel',estimator=mean, saturation=0.5)
plt.show
```
