## 1-tuple
- ordered collection of elements
- enclosed in () round braces/parenthensis
- different kind of element can be stored
- once elements can stored you can not change them(unmutateable)
  

```python
tup1=(1, "python", True,2.5)
tup1
```

```python
# type of tuple
type(tup1)
```

**- indexing in tuple**

```python
tup1[1]
```

```python
tup1[0]
```

```python
#last element is exclusive
tup1[0:3]
```

```python
# counts of elements in tuple
len(tup1)
```

```python
tup2=(2, "papa", 3.5, "false")
tup2
```

```python
# concatinate when we add tuple
tup1 + tup2
```

```python
#concatinate + repeat
tup1*2 + tup2
```

```python
tup3=(20, 30,40,50)
tup3
```

```python
min(tup3)
```

```python
max(tup3)
```

```python
tup3*2
```

- ordered collection of elements
- enclosed in square braces[]
- mutateable

```python
list1=[2, "papa", 3.5]
list1
```

```python
type(list1)
```

```python
len(list1)
```

```python
list1[2]
```

```python
list2=[3,5,"uzair","what",555,8.98,"true"]
list2
```

```python
list1 + list2
```

```python
list1*2 + list2
```

```python
list1.reverse()
list1
```

```python
list2.append("what the hell it will take to much time")
list2
```

```python
list3=[20,45,67,98,90,99]
list3
```

```python
len(list3)
```

```python
list3.sort()
list3
```

### 3- dictionaries
- an unordered collectiin of elements
- key and value
- curly braces/ or brackets
- mutateable
- 

```python
# food and thier prices
food1 = {"samosa":30, "pakora":100,"raita":10, "salad":40}
food1
```

```python
# extract food
keys1 = food1.keys()
keys1
```

```python
value1 = food1.values()
value1
```

```python
food1["Tikki"]=10
food1

```

```python
# update the value
food1["Tikki"]=15
food1
```

```python
food2 = {"dates": 50, "chocolates": 200, "sawwaya": 1000}
food2
```

```python
#concatinate
food1.update(food2)
```

```python
food1
```
